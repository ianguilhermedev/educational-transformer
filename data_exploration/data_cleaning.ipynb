```python
import pandas as pd
from pathlib import Path
import sys
```

```python
ROOT_PATH = Path().resolve().parent

if ROOT_PATH not in sys.path:
    sys.path.append(ROOT_PATH)
```

```python
data_path = ROOT_PATH / "data/raw/tatoeba_data/Pares%20de%20frase%20em%20Ingl%C3%AAstraduzidas%20para%20o%20Portugu%C3%AAs-%20-%202026-10-23.tsv"

df = pd.read_csv(data_path, sep="\t")
```

```python
#moficando nome das colunas
column_names = ['en_ids', 'en_sentences', 'pt_ids', "pt_sentences"]
df.columns = column_names

#removendo coluna dos ids
remove_columns = ['en_ids', 'pt_ids']

df.drop(
    remove_columns,
    axis=1,
    inplace=True
)
```

```python
save_data_path = ROOT_PATH / "data/processed/tatoeba_data/en-pt-pairs.tsv"

#if not save_data_path.exists():
    #save_data_path.parent.mkdir(parents=True)

df.to_csv(
    save_data_path,
    sep="\t",
    index=False
)
```
